```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

## Welcome to the World Happiness Analysis Program!
```

```python
# === 1. Load Dataset === 
df = pd.read_csv(r"happiness.csv")
print("Data loaded successfully.")
```

```python
# === 2. Explore Dataset ===

print("DISPLAY FIRST 5 ROWS:\n")
print(df.head(5).to_string())
### to_string() is a built-in Pandas DataFrame method used to convert a DataFrame into a readable string (text) representation

print("\nSHAPE:")
print(df.shape)

print("\nCOLUMNS:")
print(df.columns.to_list())

print("\nDATA TYPES:")
print(df.dtypes)
```

```python
from IPython.display import display
### The display() function from IPython is used to show a DataFrame as a clean, formatted table in Jupyter Notebook. print() shows the same data as plain text.”

# === 3. Data Cleaning ===
print("========== DATA CLEANING ==========")

##  Check missing values
print("\n1. MISSING VALUES")
missing_values = df.isnull().sum()
display(missing_values.to_frame(name="Missing Count"))
df = df.dropna()

## Check and remove duplicates
print("\n2. DUPLICATE VALUES")
print("Total Duplicates:", df.duplicated().sum())
df = df.drop_duplicates()

## Remove extra spaces from text columns
print("\n3. REMOVING EXTRA SPACES")
text_cols = df.select_dtypes(include=["object", "string"]).columns

for col in text_cols:
    df[col] = df[col].str.strip()

print("Extra spaces removed successfully!")

print("\nData Cleaning Completed Successfully!")
```

```python
from IPython.display import display
# === 4. Data Analysis===

## Top 10 Happiest Countries
top10 = df.nlargest(10, "Score")

print("\nTOP 10 HAPPIEST COUNTRIES")
display(top10[["Country or region", "Score"]].reset_index(drop=True))

## Bottom 10 Least Happy Countries
bottom10 = df.nsmallest(10, "Score")

print("\nBOTTOM 10 LEAST HAPPY COUNTRIES")
display(bottom10[["Country or region", "Score"]].reset_index(drop=True))

## === Statistical Analysis ===

columns = [
    "GDP per capita",
    "Social support",
    "Healthy life expectancy"
]
statistics = df[columns].describe().T
display(statistics)

print("\nStatistical Analysis Completed Successfully!")
```

```python
# === 5. Correlation Analysis ===

print("GDP:", round(df["Score"].corr(df["GDP per capita"]), 3))
print("Social Support:", round(df["Score"].corr(df["Social support"]), 3))
print("Life Expectancy:", round(df["Score"].corr(df["Healthy life expectancy"]), 3))
print("\nCorrelation Analysis Completed Successfully!")
```

```python
# Bar chart

plt.figure(figsize=(10, 5))
plt.bar(top10["Country or region"], top10["Score"], color="skyblue")
plt.title("Top 10 Happiest Countries", fontsize=18)
plt.xlabel("Country or Region", fontsize=14)
plt.ylabel("Score", fontsize=14)
# Increase X-axis and Y-axis label size
plt.xticks(rotation=50, ha="right", fontsize=12)
plt.yticks(fontsize=12)
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 5))
sns.barplot(data=bottom10, x="Country or region", y="Score" ,color="pink")
plt.title("Top 10 Least Happy Countries", fontsize=18)
plt.xlabel("Country or Region", fontsize=14)
plt.ylabel("Happiness Score", fontsize=14)
plt.xticks(fontsize=10, rotation = 90)
plt.yticks(fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# Scatter Plot

sns.scatterplot(data=df, x="GDP per capita", y="Score")
plt.title("GDP vs Happiness Score")
plt.show()

sns.scatterplot(data=df, x="Social support", y="Score")
plt.title("Social Support vs Happiness Score")
plt.show()

sns.scatterplot(data=df, x="Healthy life expectancy", y="Score")
plt.title("Life Expectancy vs Happiness Score")
plt.show()
```

```python
# Heatmap

cols = ["Score", "GDP per capita", "Social support", "Healthy life expectancy"]
sns.heatmap(df[cols].corr(), annot=True, cmap="coolwarm")
plt.xticks(rotation =90)
plt.title("Correlation Heatmap")
plt.show()
```
